## Loops

### While

```python
i = 0
while(i < 5):
    print(i)
    i += 1
```

```python
i = 0
while(i < 10):
    print( 2 * i + 3 )
    i += 3
```

### For

```python
for i in range(5):
    print(i)
```

```python
cars = ['Ford', 'Chevy', 'Toyota', 'Honda', 'Nissan']
```

```python
for car in cars:
    print(car)
```

```python
course = "Business Context in Data Analytics"
```

```python
all_letters = []
for letter in course:
    all_letters.append(letter)
```

```python
print(all_letters)
```

```python
print(course.lower())
print(course.upper())
```

```python
all_letters_2 = [letter for letter in course]
print(all_letters_2)
```

```python
vowel_letters = 'aeiouAEIOU'
```

```python
course_vowel_letters = [letter for letter in course if letter in vowel_letters]
print(course_vowel_letters)
```

```python
course_non_verbal_letters = [letter for letter in course if letter not in vowel_letters]
print(course_non_verbal_letters)
```

```python
vowel_letters_low = 'aeiou'
```

```python
course_vowel_letters_2 = [letter for letter in course.lower() if letter in vowel_letters_low]
print(course_vowel_letters_2)
```

```python
course_non_vowel_letters_2 = [letter for letter in course.lower() if letter not in vowel_letters_low]
print(course_non_vowel_letters_2)
```
