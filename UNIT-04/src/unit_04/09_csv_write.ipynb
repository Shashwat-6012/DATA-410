```python
import csv
```

```python
makes = ['Ford', 'Chevy', 'Toyota', 'Honda']
models = ['F-150', 'Camaro', 'Corolla', 'Civic']
max_speeds = [120, 155, 180, 130]
power = [290, 455, 139, 158]
```

```python
cars = list(zip(makes, models, max_speeds, power))
cars
```

```python
with open('../../datasets/cars.csv', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Make', 'Model', 'Max Speed', 'Power'])
    writer.writerows(cars)
```
