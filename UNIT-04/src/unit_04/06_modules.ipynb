```python
import random as rn
```

```python
random_number = rn.randint(1, 100)
random_float = rn.uniform(1.0, 10.0)

print(f"Random integer: {random_number}")
print(f"Random float: {random_float:.2f}")
```

```python
random_list = rn.sample(range(1, 100), 7)
random_list
print(f"Random list of 7 unique numbers: {random_list}")
```
