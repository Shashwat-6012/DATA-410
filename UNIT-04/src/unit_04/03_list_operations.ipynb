```python
values = [20, 27, 29, 70, 99, 83]
```

```python
cars = [ "Ford", "Chevy", "Toyota", "Honda", "Nissan"]
```

```python
print("Number of values:", len(values))
print("Number of cars:", len(cars))
```

```python
print(f"{min(values)} is the minimum value in the list of values")
print(f"{max(values)} is the maximum value in the list of values")
```

```python
avg = sum(values) / len(values)
print(f"{avg} is the average value in the list of values")
```
